```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
df = pd.read_csv("penguins.csv")
```

```python
df.head(3)
```

```python
df["bill_length_mm"] = df["bill_length_mm"].fillna(df["bill_length_mm"].mean())
df["bill_depth_mm"] = df["bill_depth_mm"].fillna(df["bill_depth_mm"].mean())

x = df["bill_length_mm"].values
y = df["bill_depth_mm"].values

print(x)
print(y)
```

```python
plt.scatter(x, y)
plt.show()
```

```python
data = list(zip(x, y))
```

```python
kmeans = KMeans(n_clusters=3)
result = kmeans.fit(data)
plt.scatter(x, y, c=kmeans.labels_)
plt.show()
```

```python
centroids  = kmeans.cluster_centers_ 
plt.scatter(x, y, c=kmeans.labels_)
plt.scatter(centroids[:,0], centroids[:,1], c="red")
plt.show()
```

```python
kmeans = KMeans(n_clusters=3)
kmeans.fit(data)
plt.scatter(x, y, c=df["species"].astype("category").cat.codes.values)
plt.show()
```

```python
from sklearn.metrics import confusion_matrix

labels = df["species"].astype("category").cat.codes.values

cm = confusion_matrix(labels, kmeans.labels_)
print(cm)
```
